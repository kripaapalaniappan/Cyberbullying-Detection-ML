```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

model_path = r"C:\Users\kripa\Desktop\Cyberbullying_Project\xlm_roberta_cyberbullying"

tokenizer = AutoTokenizer.from_pretrained(model_path)
model = AutoModelForSequenceClassification.from_pretrained(model_path)

model.eval()

print("Model Loaded Successfully!")
```

```python
label_names = {
    0: "Identity_Hate",
    1: "Normal",
    2: "Offensive",
    3: "Targeted_Individual",
    4: "Threat"
}
```

```python
import torch

def predict_comment(text):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        padding=True,
        max_length=128
    )

    with torch.no_grad():
        outputs = model(**inputs)

    predicted_label = torch.argmax(outputs.logits, dim=1).item()

    print("Comment:", text)
    print("Predicted Label:", predicted_label)
    print("Category:", label_names[predicted_label])
```

```python
predict_comment("Have a wonderful day!")
```

```python
predict_comment("You are so stupid.")
```

```python
predict_comment("I will kill you.")
```

```python
predict_comment("People like you don't belong here.")
```

```python
comments = [
    "Have a wonderful day!",
    "You are stupid.",
    "I will kill you.",
    "People like you don't belong here.",
    "Thank you for your help!"
]

for comment in comments:
    predict_comment(comment)
    print("-" * 50)
```
