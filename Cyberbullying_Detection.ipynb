```python
print("Missing/empty text:")

empty_text = (
    df_cyber["text"].isna() |
    (df_cyber["text"].astype(str).str.strip() == "")
)

print("Empty texts:", empty_text.sum())
```

```python
# Check whether duplicate texts have conflicting labels

duplicate_texts = df_cyber[
    df_cyber["text"].duplicated(keep=False)
].sort_values("text")

conflicting_duplicates = (
    duplicate_texts
    .groupby("text")["final_label"]
    .nunique()
)

conflicting_duplicates = conflicting_duplicates[
    conflicting_duplicates > 1
]

print("Duplicate text records:", len(duplicate_texts))
print("Unique duplicated texts:", duplicate_texts["text"].nunique())
print("Texts with conflicting labels:", len(conflicting_duplicates))
```

```python
# Show examples of conflicting duplicate texts

if len(conflicting_duplicates) > 0:
    for text in conflicting_duplicates.index[:10]:
        print("\nTEXT:")
        print(text)
        print("\nLabels:")
        print(
            duplicate_texts[
                duplicate_texts["text"] == text
            ][["final_label", "lang"]]
        )
else:
    print("No conflicting duplicate labels found.")
```

```python
# Create a copy so the original dataset is never modified
df_clean = df_cyber.copy()

# Remove the temporary column we created earlier
if "text_length" in df_clean.columns:
    df_clean = df_clean.drop(columns=["text_length"])

print("Original records:", len(df_clean))
```

```python
# Remove missing or empty text
df_clean = df_clean.dropna(subset=["text", "final_label"])

df_clean = df_clean[
    df_clean["text"].astype(str).str.strip() != ""
]

print("After removing empty text:", len(df_clean))
```

```python
# Find texts that have more than one label
label_counts = df_clean.groupby("text")["final_label"].nunique()

conflicting_texts = label_counts[label_counts > 1].index

print("Conflicting unique texts:", len(conflicting_texts))

# Remove all occurrences of those conflicting texts
df_clean = df_clean[
    ~df_clean["text"].isin(conflicting_texts)
]

print("After removing conflicting texts:", len(df_clean))
```

```python
before = len(df_clean)

df_clean = df_clean.drop_duplicates(
    subset=["text"],
    keep="first"
).reset_index(drop=True)

removed = before - len(df_clean)

print("Duplicate rows removed:", removed)
print("Final cleaned records:", len(df_clean))
```

```python
print("Final label distribution:")
print(df_clean["final_label"].value_counts())

print("\nFinal label percentages:")
print(
    df_clean["final_label"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# Keep only English and Hindi records

df_model = df_clean[
    df_clean["lang"].isin(["en", "hi"])
].copy().reset_index(drop=True)

print("Records for model:", len(df_model))

print("\nLanguage distribution:")
print(df_model["lang"].value_counts())

print("\nLabel distribution:")
print(df_model["final_label"].value_counts())
```

```python
from datasets import Dataset

# Keep only text and label
train_data = train_df[["text", "final_label"]].copy()
val_data = val_df[["text", "final_label"]].copy()
test_data = test_df[["text", "final_label"]].copy()

# Rename label column
train_data = train_data.rename(columns={"final_label": "labels"})
val_data = val_data.rename(columns={"final_label": "labels"})
test_data = test_data.rename(columns={"final_label": "labels"})

# Convert to Hugging Face Dataset
train_dataset = Dataset.from_pandas(
    train_data,
    preserve_index=False
)

val_dataset = Dataset.from_pandas(
    val_data,
    preserve_index=False
)

test_dataset = Dataset.from_pandas(
    test_data,
    preserve_index=False
)

print("Hugging Face datasets created!")
print("Train:", len(train_dataset))
print("Validation:", len(val_dataset))
print("Test:", len(test_dataset))
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification

MODEL_NAME = "xlm-roberta-base"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=2
)

print("XLM-RoBERTa tokenizer and model loaded!")
print("Model:", MODEL_NAME)
print("Labels:", model.config.num_labels)
```

```python
def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        truncation=True,
        max_length=128
    )

train_tokenized = train_dataset.map(
    tokenize_function,
    batched=True
)

val_tokenized = val_dataset.map(
    tokenize_function,
    batched=True
)

test_tokenized = test_dataset.map(
    tokenize_function,
    batched=True
)

print("Tokenization complete!")
print("Train:", len(train_tokenized))
print("Validation:", len(val_tokenized))
print("Test:", len(test_tokenized))
```

```python
# Reset the dataset format
train_tokenized.set_format(type=None)
val_tokenized.set_format(type=None)
test_tokenized.set_format(type=None)

print("Datasets ready for training!")

print("\nTrain columns:")
print(train_tokenized.column_names)

print("\nExample:")
print(train_tokenized[0])
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="./cyberbullying_xlmr_model",

    # Training
    num_train_epochs=2,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,

    # Evaluation
    eval_strategy="epoch",
    save_strategy="epoch",

    # Learning
    learning_rate=2e-5,
    weight_decay=0.01,

    # Performance
    fp16=True,

    # Keep the best model
    load_best_model_at_end=True,

    # Logging
    logging_steps=100,
    report_to="none"
)

print("Training configuration created successfully!")
```

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_tokenized,
    eval_dataset=val_tokenized
)

print("Trainer created successfully!")
```
